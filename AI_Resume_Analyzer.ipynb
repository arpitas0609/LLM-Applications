```python
import PyPDF2
def read_resume(file_path):
  text= ""
  with open(file_path, "rb") as file:
    reader = PyPDF2.PdfReader(file)
    for page in reader.pages:
      text += page.extract_text()
      return text
      print(read_resume(file_name))
```

```python
from google import genai
from google.colab import userdata
import os

try:
    # Try to load from Colab Secrets
    api_key = userdata.get('GOOGLE_API_KEY')
except Exception as e:
    # If not found, prompt for manual input or use environment variable
    print("GOOGLE_API_KEY secret not found in Colab Secrets.")
    print("Please add it to the Secrets (key icon) on the left panel, or paste it below:")
    api_key = input("Enter your Gemini API Key: ").strip()

client = genai.Client(api_key=api_key)
```

```python
def analyze_resume(text):
  response = client.models.generate_content(
      model="gemini-3.5-flash-lite",
      contents=f"""
You are an ATS resume analyzer.

Analyze the resume below and return ATS score and feedback.

Resume:
{text}

Return in JSON format with:
- ats_score (0-100)
- summary
- strengths
- weaknesses

Rules:
- Output ONLY JSON
- No extra text
"""
   )
  return response.text
```

```python
resume_text = read_resume(file_name)
result = analyze_resume(resume_text)
print(result)
```
